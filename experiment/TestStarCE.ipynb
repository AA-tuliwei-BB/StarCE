### Init

```python
import sys
import os
from pathlib import Path

# Add experiment directory to path
sys.path.append(os.path.abspath('.'))

from ExperimentRunner import StarCETestRunner

# Project root directory
project_root = os.path.abspath('..')

# Output to experiment/checkpoint/StarCE directory
checkpoint_dir = os.path.abspath('./checkpoint/StarCE')
os.makedirs(checkpoint_dir, exist_ok=True)

# Create runner
runner = StarCETestRunner(project_root)

def build_starce_benchmark(db_name):
    """
    Build StarCE statistics object

    Args:
        db_name: Database name ('STATS', 'IMDB', 'IMDB_LIGHT', 'IMDB_LIGHT_RANGES', 'JobJoin', 'StatsJoin')

    Returns:
        tuple: (build time, statistics raw data size)
    """
    if db_name == 'STATS':
        config = runner.get_stats_db_config()
    elif db_name == 'IMDB':
        config = runner.get_imdb_db_config()
    elif db_name == 'IMDB_LIGHT':
        config = runner.get_imdb_light_db_config()
    elif db_name == 'IMDB_LIGHT_RANGES':
        config = runner.get_imdb_light_ranges_db_config()
    elif db_name == 'JobJoin':
        config = runner.get_jobjoin_config()
    elif db_name == 'StatsJoin':
        config = runner.get_statsjoin_config()
    else:
        raise ValueError(f"Unsupported database name: {db_name}")

    config.RefreshStatistics = 1
    # JobJoin and StatsJoin contain large multi-table joins;
    # build only needs statistics collection, skip query execution.
    if db_name in ('JobJoin', 'StatsJoin'):
        config.SQL_PATH = 'dummy_query.sql'
    time, size = runner.inner_test_build_time(config, db_name, num_runs=1)
    print(f"\nBuild results:")
    print(f"  Build time: {time:.2f} seconds")
    print(f"  Statistics size (data_only): {size} bytes ({size/1024/1024:.2f} MB)")
    return time, size

print("Configuration complete, use build_starce_benchmark() to build each database")
```

```python
# Save build/statistics/evaluation time to CSV
import csv

benchmarks = [
    {"name": "STATS", "build_db": "STATS", "eval_benchmark": "Stats"},
    {"name": "JOBM", "build_db": "IMDB", "eval_benchmark": "JOBM"},
    {"name": "JOBLight", "build_db": "IMDB_LIGHT", "eval_benchmark": "JOBLight"},
    {"name": "JOBLightRanges", "build_db": "IMDB_LIGHT_RANGES", "eval_benchmark": "JOBLightRanges"},
    {"name": "JobJoin", "build_db": "JobJoin", "eval_benchmark": "JobJoin"},
    {"name": "StatsJoin", "build_db": "StatsJoin", "eval_benchmark": "StatsJoin"},
]

rows = []
for item in benchmarks:
    name = item["name"]

    if name == "STATS":
        build_time, stats_size = stats_build_time, stats_stats_size
        eval_result = stats_results
    elif name == "JOBM":
        build_time, stats_size = jobm_build_time, jobm_stats_size
        eval_result = jobm_results
    elif name == "JOBLight":
        build_time, stats_size = joblight_build_time, joblight_stats_size
        eval_result = joblight_results
    elif name == "JOBLightRanges":
        build_time, stats_size = joblr_build_time, joblr_stats_size
        eval_result = joblr_results
    elif name == "JobJoin":
        build_time, stats_size = jobjoin_build_time, jobjoin_stats_size
        eval_result = jobjoin_results
    elif name == "StatsJoin":
        build_time, stats_size = statsjoin_build_time, statsjoin_stats_size
        eval_result = statsjoin_results

    rows.append({
        "benchmark": name,
        "build_time_sec": build_time,
        "statistics_size_bytes": stats_size,
        "eval_time_sec": eval_result.get("total_time"),
    })

csv_path = os.path.join(checkpoint_dir, "benchmark_times.csv")
with open(csv_path, "w", newline="") as f:
    writer = csv.DictWriter(
        f,
        fieldnames=["benchmark", "build_time_sec", "statistics_size_bytes", "eval_time_sec"],
    )
    writer.writeheader()
    writer.writerows(rows)

print(f"Written to: {csv_path}")
```
